# Partie 1

```python
temp=[24.5, 25.1, 26.3, 27.0 , 25.8]
```

```python
import numpy as np

tempNP=np.array(temp)
```

```python
print(temp*2)
print(tempNP*2)
```

On constate que l'opération de multiplication est effectuée de deux manières différentes : la liste duplique ses éléments, tandis que le ndarray multiplie chacun de ses éléments par 2.

# Partie 8 - Calcul scientifique
